## Importing necessary libraries

```python
import pandas as pd
import nltk
import re
from nltk.corpus import brown
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
```

## Importing dataset

This part was all about importing the dataset. The dataset I picked was Netflix Reviews on Kaggle with this link: https://www.kaggle.com/code/darrylljk/netflix-reviews-with-nlp.
In the dataset the one that been import are the 8 columns which cointains: ['reviewId', 'userName', 'content', 'score', 'thumbsUpCount',
       'reviewCreatedVersion', 'at', 'appVersion']
Also the shape of the dataset that chosen has a shape of : 113068

```python
#import dataset
df = pd.read_csv('netflix_reviews.csv')

print("Shape of the dataset:", df.shape)
print("Columns in the dataset:", df.columns)

#check the first 5 instances
df.head()
```

## Unitize and Tokenize

 fillna('') replaces any missing/null values in the 'content' column with empty strings to prevent errors, while  word_tokenize breaks down each review into individual words (tokens)

```python
# Handle missing values and apply word_tokenize function to each article
df['tokenized_content'] = df['content'].fillna('').apply(word_tokenize)

df.head()
```

## Lowercase

In this step from its words Lowercase: It Converts every token (word) to lowercase using a lambda function and Iterates through each token in the list and applies .lower()

```python
# Convert tokens to lowercase
df['tokenized_content'] = df['tokenized_content'].apply(lambda x: [token.lower() for token in x])

df.head()
```

We can see here this step is to make uniformity and reduces duplication and decrease unique word count example: "Netflix", "netflix", and "NETFLIX" as the same word

## Remove Digits

In this step we are Removing any token that contains at least one digit (0-9) with the function any(char.isdigit() for char in token) to check each character.

```python
df['tokenized_content'] = df['tokenized_content'].apply(lambda x: [token.lower() for token in x if not any(char.isdigit() for char in token)])
df.head()
```

## Remove Special Characters and White Spaces

On this step we are to replaces matched characters with empty string (import re). And  if token removes empty strings (tokens that were only punctuation) 

 Before: ['i', "'m", ',', 'so', 'far', 'smooth', 'sailing.app']
 Now: ['i', 'm', 'so', 'far', 'smooth', 'sailingapp']

 So this essentially removed punctuation, special symbols and whitespaces.

```python
import re

# Remove special characters from tokens
df['tokenized_content'] = df['tokenized_content'].apply(
    lambda x: [token for token in 
               (re.sub(r'[^a-zA-Z0-9]', '', t) for t in x)
               if token]
)
df.head()
```

## Remove Stop Words

```python
#download and import english stopwords
stop_words = stopwords.words('english')
print(stop_words)
```

```python
df['tokenized_content'] = df['tokenized_content'].apply(
    lambda x: [token for token in x if token.lower() not in stop_words]
)

df.head()
```

## Stemming

In this step we are in the process of reducing words to their root or base form by removing suffixes. It uses algorithmic rules (not a dictionary) to chop off word endings.

## Remove Stop Words

Example: 

Original              Stemmed                               Explanation
forced                 forc                                  Removes ed

```python
from nltk.stem import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Apply stemming to each token
df['stemmed_content'] = df['tokenized_content'].apply(lambda x: [stemmer.stem(token) for token in x])

df.head()
```

## Lemmatization

Lastly we are to lemmatize word to their dictionary base form (lemma) using vocabulary and morphological analysis. Unlike stemming, it produces real, valid words.

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')

# Initialize the WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Apply lemmatization to each token
df['lemmatized_content'] = df['tokenized_content'].apply(lambda x: [lemmatizer.lemmatize(token) for token in x])

df.head()
```
